````python
# 5.1. Пасажири, які вижили

survived = titanic[titanic['survived'] == 1]
total = len(titanic)

print(f"5.1. Вижило: {len(survived)}")
print(f"5.1. Відсоток: {len(survived) / total * 100:.2f}%")

# 5.2. Жінки 1 класу, які вижили

female_1st_survived = titanic[
    (titanic['sex'] == 'female') &
    (titanic['pclass'] == 1) &
    (titanic['survived'] == 1)
]

print(f"\n5.2. Кількість: {len(female_1st_survived)}")
print(female_1st_survived.head())

# 5.3. Діти (<12) або 1 клас

filtered = titanic[
    (titanic['age'] < 12) |
    (titanic['pclass'] == 1)
]

survived_percent = filtered['survived'].mean() * 100

print(f"\n5.3. Кількість: {len(filtered)}")
print(f"5.3. Відсоток виживших: {survived_percent:.2f}%")

# 5.4. Через query + середній вік

query_filtered = titanic.query(
    "(sex == 'female') & (pclass == 1) & (survived == 1)"
)

avg_age = titanic['age'].mean()

above_avg_age = titanic.query("age > @avg_age")

print(f"\n5.4. Через query (жінки 1 класу, які вижили): {len(query_filtered)}")
print(f"5.4. Старше середнього віку: {len(above_avg_age)}")

# 5.5. Cherbourg або Queenstown vs Southampton

cq = titanic[titanic['embark_town'].isin(['Cherbourg', 'Queenstown'])]
southampton = titanic[titanic['embark_town'] == 'Southampton']

print(f"\n5.5. Середня ціна (Cherbourg + Queenstown): {cq['fare'].mean():.2f}")
print(f"5.5. Середня ціна (Southampton): {southampton['fare'].mean():.2f}")

# 5.6. Вік 20–30 + топ 10 дорогих

age_20_30 = titanic[(titanic['age'] >= 20) & (titanic['age'] <= 30)]

survived_20_30 = age_20_30['survived'].sum()

print(f"\n5.6. Пасажири 20–30 років: {len(age_20_30)}")
print(f"5.6. З них вижило: {int(survived_20_30)}")

top10_fare = titanic.sort_values(by='fare', ascending=False).head(10)

cols_exist = [c for c in ['name', 'pclass', 'fare', 'survived'] if c in titanic.columns]

print("\n5.6. Топ-10 найдорожчих квитків:")
print(top10_fare[cols_exist])
````

## Завдання 6. Обробка пропусків та трансформація

### Завдання:

**6.1. Діагностика пропусків.**
- Підрахуйте кількість пропущених значень у кожному стовпці.
- Обчисліть **відсоток** пропусків для кожного стовпця.
- Визначте, які стовпці мають критично високий рівень пропусків (>50%).

**6.2. Стратегія обробки.** Працюйте з копією: `titanic_clean = titanic.copy()`.
- Стовпець `age` — заповніть **медіаною**.
- Стовпець `embarked` — заповніть **модою** (найчастішим значенням).
- Стовпець `embark_town` — заповніть відповідно до `embarked`.
- Стовпець `deck` — огляньте дані колонки. Прийміть рішення, що робити з такою колонкою
- Перевірте, що пропусків у ключових стовпцях не залишилось.

**6.3. Створення нових стовпців.**
- ціна на одного члена групи подорожуючих, де +1 — сам пасажир
- `age_group` — категорія віку: `'дитина'` (0–12), `'підліток'` (13–17), `'дорослий'` (18–59), `'літній'` (60+)

**6.4. Сортування.**
- Відсортуйте `titanic_clean` за `fare` за спаданням. Виведіть топ-5 записів.
- Відсортуйте за двома стовпцями: `pclass` (зростання) та `age` (спадання). Виведіть перші 10.

**6.5. Перейменування та видалення.**
- Перейменуйте стовпці: `survived` → `вижив`, `pclass` → `клас`, `sex` → `стать`, `age` → `вік`.
- Видаліть стовпці `alive` та `who` (вони дублюють інформацію з інших стовпців). Поясніть, чому дублювання стовпців — проблема.

````python
# 6.1. Діагностика пропусків

missing_count = titanic.isnull().sum()
missing_percent = titanic.isnull().mean() * 100

print("6.1. Кількість пропусків у кожному стовпці:")
print(missing_count)

print("\n6.1. Відсоток пропусків у кожному стовпці:")
print(missing_percent)

high_missing = missing_percent[missing_percent > 50]

print("\n6.1. Стовпці з високим рівнем пропусків (>50%):")
print(high_missing)

# 6.2. Обробка пропусків

titanic_clean = titanic.copy()

titanic_clean['age'] = titanic_clean['age'].fillna(titanic_clean['age'].median())
titanic_clean['embarked'] = titanic_clean['embarked'].fillna(titanic_clean['embarked'].mode()[0])

embarked_to_town = (
    titanic_clean[['embarked', 'embark_town']]
    .dropna()
    .drop_duplicates()
    .set_index('embarked')['embark_town']
    .to_dict()
)

titanic_clean['embark_town'] = titanic_clean['embark_town'].fillna(
    titanic_clean['embarked'].map(embarked_to_town)
)

titanic_clean = titanic_clean.drop(columns=['deck'])

print("\n6.2. Пропуски після обробки:")
print(titanic_clean[['age', 'embarked', 'embark_town']].isnull().sum())

# 6.3. Створення нових стовпців

titanic_clean['fare_per_person'] = titanic_clean['fare'] / (titanic_clean['sibsp'] + titanic_clean['parch'] + 1)

titanic_clean['age_group'] = pd.cut(
    titanic_clean['age'],
    bins=[0, 12, 17, 59, 120],
    labels=['дитина', 'підліток', 'дорослий', 'літній'],
    include_lowest=True
)

print("\n6.3. Нові стовпці fare_per_person та age_group:")
print(titanic_clean[['fare_per_person', 'age_group']].head())

# 6.4. Сортування

fare_sorted = titanic_clean.sort_values(by='fare', ascending=False)
print("\n6.4. Топ-5 за fare:")
print(fare_sorted.head())

class_age_sorted = titanic_clean.sort_values(by=['pclass', 'age'], ascending=[True, False])
print("\n6.4. Перші 10 за pclass зростанням та age спаданням:")
print(class_age_sorted.head(10))

# 6.5. Перейменування та видалення

titanic_clean = titanic_clean.rename(columns={
    'survived': 'вижив',
    'pclass': 'клас',
    'sex': 'стать',
    'age': 'вік'
})

cols_to_drop = [col for col in ['alive', 'who'] if col in titanic_clean.columns]
titanic_clean = titanic_clean.drop(columns=cols_to_drop)

print("\n6.5. Стовпці після перейменування та видалення:")
print(titanic_clean.columns)
````

````python
# 7. Дослідницьке питання

print("7.1. Питання:")
print("Як залежить відсоток виживання пасажирів від класу та статі?")

titanic_research = titanic.copy()

titanic_research['age'] = titanic_research['age'].fillna(titanic_research['age'].median())

survival_by_group = titanic_research.groupby(['pclass', 'sex'])['survived'].agg(['count', 'sum', 'mean'])
survival_by_group = survival_by_group.rename(columns={
    'count': 'кількість',
    'sum': 'вижило',
    'mean': 'відсоток_виживання'
})

survival_by_group['відсоток_виживання'] = survival_by_group['відсоток_виживання'] * 100

print("\n7.2. Результат дослідження:")
print(survival_by_group)

top_groups = survival_by_group.sort_values(by='відсоток_виживання', ascending=False)

print("\n7.3. Групи за спаданням відсотка виживання:")
print(top_groups)

print("\n7.4. Висновок:")
print("Найвищий відсоток виживання спостерігається серед жінок 1-го класу.")
print("Найнижчий відсоток виживання мають чоловіки 3-го класу.")
print("Отже, на виживання пасажирів суттєво впливали і стать, і клас квитка.")
print("Гіпотеза про те, що пасажири вищого класу та жінки мали більше шансів вижити, підтвердилась.")
````
